```python
import sys
import os
import torch
import zipfile
from pathlib import Path
from tqdm import tqdm
from PIL import Image

PROJECT_PATH = "/content/drive/MyDrive/Generative Models"
sys.path.insert(0, PROJECT_PATH)
os.chdir(PROJECT_PATH)

from models import get_generator
from data.dataset import get_transforms, SingleDomainDataset, denormalize

device = "cuda" if torch.cuda.is_available() else "cpu"

# Load G_BA from checkpoint (the correct one!)
checkpoint = torch.load(
    "/content/drive/MyDrive/Generative Models/checkpoints/final_checkpoint.pth",
    map_location=device
)

G_BA = get_generator("resnet", in_channels=3, out_channels=3).to(device)
G_BA.load_state_dict(checkpoint['G_BA_state_dict'])
G_BA.eval()

print("✅ Loaded G_BA (Photo -> Monet)")

# Setup
photo_dir = "/content/drive/MyDrive/Generative Models/gan-getting-started/photo_jpg"
output_dir = "/content/drive/MyDrive/Generative Models/generated_images"
zip_path = "/content/drive/MyDrive/Generative Models/images.zip"
num_images = 7500

Path(output_dir).mkdir(parents=True, exist_ok=True)

# Load photos
transform = get_transforms(256, mode="test")
dataset = SingleDomainDataset(photo_dir, transform=transform)

print(f"Generating {num_images} Monet-style images...")

# Generate images
count = 0
with torch.no_grad():
    for i in tqdm(range(min(num_images, len(dataset))), desc="Generating"):
        photo = dataset[i].unsqueeze(0).to(device)
        fake_monet = G_BA(photo)  # Using G_BA!

        # Convert to image
        img = denormalize(fake_monet[0])
        img = torch.clamp(img, 0, 1)
        img_np = img.cpu().permute(1, 2, 0).numpy()
        img_np = (img_np * 255).astype('uint8')
        pil_img = Image.fromarray(img_np)

        # Save
        img_path = os.path.join(output_dir, f"{count:05d}.jpg")
        pil_img.save(img_path, "JPEG", quality=95)
        count += 1

print(f"✅ Generated {count} images")

# Create zip
print("Creating submission zip...")
image_files = sorted([f for f in os.listdir(output_dir) if f.endswith('.jpg')])

with zipfile.ZipFile(zip_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for img_file in tqdm(image_files, desc="Zipping"):
        img_path = os.path.join(output_dir, img_file)
        zipf.write(img_path, img_file)

zip_size = os.path.getsize(zip_path) / (1024 * 1024)
print(f"✅ Created {zip_path}")
print(f"   {len(image_files)} images, {zip_size:.1f} MB")
print(f"\n🎉 Ready to submit to Kaggle!")
```
